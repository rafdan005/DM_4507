# 02a - Mengambil Data
Materi Modul 2: membaca data dari file CSV/JSON, database, API, dan cloud.

Data yang sama (Penguins) tersedia dalam tiga format di folder `dataset/`: `penguins.csv`, `penguins.json`, `penguins.db`.

## CSV

```python
import pandas as pd
import sqlite3

df_csv = pd.read_csv('../dataset/penguins.csv')
df_csv.head(3)
```

## JSON

```python
df_json = pd.read_json('../dataset/penguins.json')
df_json.head(3)
```

## Database (SQLite)

```python
con = sqlite3.connect('../dataset/penguins.db')
df_db = pd.read_sql('SELECT body_mass_g, flipper_length_mm, target FROM penguins WHERE body_mass_g > 5000 ORDER BY body_mass_g DESC', con)
con.close()
print(df_db.shape)
df_db.head()
```

## Memastikan ketiga sumber identik

```python
con = sqlite3.connect('../dataset/penguins.db')
df_sql_all = pd.read_sql('SELECT * FROM penguins', con)
con.close()
print('CSV vs JSON :', df_csv.round(6).astype(float).equals(df_json.round(6).astype(float)))
print('CSV vs DB   :', df_csv.round(6).astype(float).equals(df_sql_all.round(6).astype(float)))
```
