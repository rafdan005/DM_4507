# Tambahan - Clustering
K-Means pada dataset **Palmer Penguins** (label spesies asli dipakai hanya untuk evaluasi, bukan untuk training).

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.decomposition import PCA
from sklearn.metrics import silhouette_score, adjusted_rand_score
```

```python
df = pd.read_csv('../dataset/penguins.csv')
X = StandardScaler().fit_transform(df.drop(columns='target'))
y = df['target']
```

## Menentukan jumlah cluster (Elbow & Silhouette)

```python
ks = range(2, 9)
inertia, sil = [], []
for k in ks:
    km = KMeans(n_clusters=k, n_init=10, random_state=42).fit(X)
    inertia.append(km.inertia_); sil.append(silhouette_score(X, km.labels_))

fig, ax = plt.subplots(1, 2, figsize=(11,4))
ax[0].plot(ks, inertia, 'o-'); ax[0].set_title('Elbow (inertia)'); ax[0].set_xlabel('k')
ax[1].plot(ks, sil, 'o-'); ax[1].set_title('Silhouette score'); ax[1].set_xlabel('k')
plt.show()
```

```python
km = KMeans(n_clusters=3, n_init=10, random_state=42).fit(X)
print('Silhouette:', round(silhouette_score(X, km.labels_), 4))
print('Adjusted Rand vs label asli:', round(adjusted_rand_score(y, km.labels_), 4))
```

```python
pc = PCA(n_components=2).fit_transform(X)
plt.scatter(pc[:,0], pc[:,1], c=km.labels_, cmap='viridis')
plt.title('Hasil K-Means (proyeksi PCA)'); plt.xlabel('PC1'); plt.ylabel('PC2')
plt.show()
```
