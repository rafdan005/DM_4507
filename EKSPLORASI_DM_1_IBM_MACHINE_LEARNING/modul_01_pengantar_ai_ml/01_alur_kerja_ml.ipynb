Notebook ini mendemokan seluruh alur tersebut dalam versi paling ringkas dengan dataset **Palmer Penguins** (klasifikasi spesies penguin).

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import make_pipeline
from sklearn.metrics import accuracy_score
```

## 1-2. Ambil & cek data

```python
df = pd.read_csv('../dataset/penguins.csv')
print(df.shape, '| missing:', df.isna().sum().sum())
```

## 3. Siapkan fitur

```python
X, y = df.drop(columns='target'), df['target']
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, stratify=y, random_state=42)
```

## 4-5. Modeling & evaluasi

```python
model = make_pipeline(StandardScaler(), LogisticRegression(max_iter=1000))
model.fit(X_train, y_train)
print('Akurasi test:', round(accuracy_score(y_test, model.predict(X_test)), 4))
```
